```python
import pandas as pd
import numpy as np
import os
import re
import json
import csv
```

```python
def process_files(raw_folder_paths, output_csv):


    def extract_label(data):
        category = data.get('categories',[])
        if not category:
            return None
        
        label = category[0].lower()

        if "sport" in label:
            return "sports"
        elif "politic" in label:
            return "politics"
        else:
            return None
        
    def clean_article_text(raw_text):
        
        keywords = [
            'all rights reserved',
            'cable news network',
            'contributed to this report',
            '©',
            'the-cnn-wire'
        ]
        lines = raw_text.split('\n')
        cleaned_lines = []

        for line in lines:
            line = line.strip()

            if not line: continue
            if line.startswith('by '): continue
            if line.startswith('(') and ")" in line: continue
            if any(keyword in line.lower() for keyword in keywords): continue
        
            cleaned_lines.append(line)
        
        text = " ".join(cleaned_lines)
        text = re.sub(r"\s+"," ",text)

        return text.strip()
    
    dataset = []
    for raw_folder_path in raw_folder_paths:
        for article_name in os.listdir(raw_folder_path):
            print(article_name)
            article_path = os.path.join(raw_folder_path, article_name)
            
            with open(article_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
            
            raw_text = data.get('text','')
            label = extract_label(data)

            if not raw_text or not label:
                continue

            cleaned_text = clean_article_text(raw_text)

            if len(cleaned_text.split()) < 50:
                continue

            dataset.append((cleaned_text,label))
    
    with open(output_csv,"w",encoding='utf-8',newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["text","label"])
        writer.writerows(dataset)
            
```

```python
sports_folder = './dataset/Sports_Raw_Articles'
politics_folder = './dataset/Politics_Raw_Articles'
process_files([sports_folder,politics_folder],'./dataset/dataset.csv')
```

```python
df = pd.read_csv('./dataset/dataset.csv')
```

```python
sports_df = df[df['label'] == 'sports']
politics_df = df[df['label'] == 'politics']
```
